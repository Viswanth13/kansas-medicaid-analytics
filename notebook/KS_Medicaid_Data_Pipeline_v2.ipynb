## Bureau of Labor Statistics (BLS) API

```python
import requests
import json

BLS_URL = "https://api.bls.gov/publicAPI/v1/timeseries/data/"

payload = {
    "seriesid": ["LASST200000000000003"],
    "startyear": "2017",
    "endyear": "2026"
}

response = requests.post(
    BLS_URL,
    json=payload
)

print("HTTP Status:", response.status_code)
```

```python
bls_json = response.json()

print("BLS Status:", bls_json["status"])
print("Messages:", bls_json["message"])
```

```python
bls_json["Results"]["series"][0]["data"][:3]
```

```python
bls_records = bls_json["Results"]["series"][0]["data"]
```

```python
df_bls = pd.DataFrame(bls_records)

df_bls.head()
```

```python
df_bls = df_bls[
    ["year", "period", "periodName", "value"]
].copy()
```

```python
df_bls.head()
```

```python
df_bls = df_bls[
    df_bls["period"].str.match(r"M(0[1-9]|1[0-2])")
].copy()
```

```python
df_bls["period"].unique()
```

```python
df_bls["month"] = (
    df_bls["period"]
    .str.replace("M", "", regex=False)
    .astype(int)
)
```

```python
df_bls["Reporting Month"] = pd.to_datetime(
    dict(
        year=df_bls["year"].astype(int),
        month=df_bls["month"],
        day=1
    )
)
```

```python
df_bls["Unemployment Rate"] = pd.to_numeric(
    df_bls["value"],
    errors="coerce"
)
```

## Analysis

```python
kansas_combined[
    [
        "Reporting Month",
        "Total Medicaid and CHIP Enrollment",
        "Total Medicaid Enrollment",
        "Total CHIP Enrollment",
        "New Applications Submitted to Medicaid and CHIP Agencies",
        "Unemployment Rate Filled"
    ]
].head(10)
```

```python
applications_col = "New Applications Submitted to Medicaid and CHIP Agencies"

print(kansas_combined[applications_col].describe())

print(
    "\nNumber of zero values:",
    (kansas_combined[applications_col] == 0).sum()
)

print(
    "Number of missing values:",
    kansas_combined[applications_col].isnull().sum()
)
```

```python
kansas_combined.loc[
    kansas_combined[applications_col] > 0,
    ["Reporting Month", applications_col]
].head(10)
```

```python
start_enrollment = kansas_combined.loc[
    kansas_combined["Reporting Month"].idxmin(),
    "Total Medicaid and CHIP Enrollment"
]

end_enrollment = kansas_combined.loc[
    kansas_combined["Reporting Month"].idxmax(),
    "Total Medicaid and CHIP Enrollment"
]

overall_change = end_enrollment - start_enrollment

overall_change_pct = (
    overall_change / start_enrollment
) * 100

print("Start enrollment:", start_enrollment)
print("End enrollment:", end_enrollment)
print("Absolute change:", overall_change)
print("Percent change:", round(overall_change_pct, 2), "%")
```

```python
max_row = kansas_combined.loc[
    kansas_combined["Total Medicaid and CHIP Enrollment"].idxmax()
]

min_row = kansas_combined.loc[
    kansas_combined["Total Medicaid and CHIP Enrollment"].idxmin()
]

print(
    "Highest enrollment:",
    int(max_row["Total Medicaid and CHIP Enrollment"]),
    "in",
    max_row["Reporting Month"].date()
)

print(
    "Lowest enrollment:",
    int(min_row["Total Medicaid and CHIP Enrollment"]),
    "in",
    min_row["Reporting Month"].date()
)
```

```python
increase_from_min_to_max = (
    max_row["Total Medicaid and CHIP Enrollment"]
    - min_row["Total Medicaid and CHIP Enrollment"]
)

increase_pct = (
    increase_from_min_to_max
    / min_row["Total Medicaid and CHIP Enrollment"]
) * 100

print("Increase in enrollment:", int(increase_from_min_to_max))
print("Percent increase:", round(increase_pct, 2), "%")
```

```python
latest_row = kansas_combined.loc[
    kansas_combined["Reporting Month"].idxmax()
]

decline_from_peak = (
    latest_row["Total Medicaid and CHIP Enrollment"]
    - max_row["Total Medicaid and CHIP Enrollment"]
)

decline_pct = (
    decline_from_peak
    / max_row["Total Medicaid and CHIP Enrollment"]
) * 100

print("Latest enrollment:", int(latest_row["Total Medicaid and CHIP Enrollment"]))
print("Change from peak:", int(decline_from_peak))
print("Percent change from peak:", round(decline_pct, 2), "%")
```

```python
kansas_combined["Enrollment YoY Change %"] = (
    kansas_combined["Total Medicaid and CHIP Enrollment"]
    .pct_change(periods=12) * 100
)

kansas_combined[
    [
        "Reporting Month",
        "Total Medicaid and CHIP Enrollment",
        "Enrollment YoY Change %"
    ]
].tail(20)
```

```python
max_yoy_row = kansas_combined.loc[
    kansas_combined["Enrollment YoY Change %"].idxmax()
]

min_yoy_row = kansas_combined.loc[
    kansas_combined["Enrollment YoY Change %"].idxmin()
]

print(
    "Largest YoY increase:",
    round(max_yoy_row["Enrollment YoY Change %"], 2), "%",
    "in",
    max_yoy_row["Reporting Month"].date()
)

print(
    "Largest YoY decline:",
    round(min_yoy_row["Enrollment YoY Change %"], 2), "%",
    "in",
    min_yoy_row["Reporting Month"].date()
)
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    kansas_combined["Reporting Month"],
    kansas_combined["Enrollment YoY Change %"]
)

plt.axhline(
    0,
    linestyle="--"
)

plt.axvline(
    pd.Timestamp("2020-03-01"),
    linestyle="--"
)

plt.axvline(
    pd.Timestamp("2023-04-01"),
    linestyle="--"
)

plt.title("Kansas Medicaid and CHIP Enrollment - Year-over-Year Change")
plt.xlabel("Reporting Month")
plt.ylabel("YoY Change (%)")
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
pre_period = kansas_combined[
    kansas_combined["Reporting Month"] < "2020-03-01"
]

continuous_period = kansas_combined[
    (kansas_combined["Reporting Month"] >= "2020-03-01") &
    (kansas_combined["Reporting Month"] < "2023-04-01")
]

post_period = kansas_combined[
    kansas_combined["Reporting Month"] >= "2023-04-01"
]

print(
    "Pre-period average YoY:",
    round(pre_period["Enrollment YoY Change %"].mean(), 2), "%"
)

print(
    "Continuous-enrollment period average YoY:",
    round(continuous_period["Enrollment YoY Change %"].mean(), 2), "%"
)

print(
    "Post-April-2023 average YoY:",
    round(post_period["Enrollment YoY Change %"].mean(), 2), "%"
)
```
